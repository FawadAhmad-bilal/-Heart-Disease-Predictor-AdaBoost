```python
df["trestbps"]=df["trestbps"].fillna(df["trestbps"].median())
df["chol"]=df["chol"].fillna(df["chol"].median())
df["fbs"]=df["fbs"].fillna(df["fbs"].mode()[0])
df["restecg"]=df["restecg"].fillna(df["restecg"].mode()[0])
df["exang"]=df["exang"].fillna(df["exang"].mode()[0])
df["oldpeak"]=df["oldpeak"].fillna(df["oldpeak"].median())
df["thalch"]=df["thalch"].fillna(df["thalch"].median())
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df["cp"].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df["restecg"]=le.fit_transform(df["restecg"])
df["sex"]=le.fit_transform(df["sex"])
df["fbs"]=le.fit_transform(df["fbs"])
df["exang"]=le.fit_transform(df["exang"])
df = pd.get_dummies(df, columns=['cp'], drop_first=True)
```

```python
df.head()
```

```python
df["chol"]=df["chol"].replace(0,df["chol"].median())
df["trestbps"]=df["trestbps"].replace(0,df["trestbps"].median())
```

```python
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,mean_squared_error,confusion_matrix,classification_report
```

```python
y = (df['num'] > 0).astype(int)   
x = df.drop(columns=['num'])  
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2,stratify=y)
```

```python
dt=DecisionTreeClassifier(max_depth=1)
```

```python
dt.fit(x_train,y_train)
```

```python
pre=dt.predict(x_test)
```

```python
accuracy_score(y_test,pre)
```

```python
AdaBoostClassifier().get_params()
```

```python
adb=AdaBoostClassifier(
    estimator=dt,
    n_estimators=200,
    learning_rate=0.1,
    random_state=42)
```

```python
adb.fit(x_train,y_train)
```

```python
pre_y=adb.predict(x_test)
```

```python
accuracy_score(y_test,pre_y)
```

```python
cm=confusion_matrix(y_test,pre_y)
```

```python
sns.heatmap(cm,annot=True,linecolor='white',xticklabels=['No Disease', 'Disease'],
    yticklabels=['No Disease', 'Disease'])
    # annot_kws={'size': 16, 'weight': 'bold'})
```

```python
classification_report(y_test,pre_y)
```
